```python
import requests
from PyPDF2 import PdfReader
from io import BytesIO

# 1. Extract text from PDF
urls = [
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0027.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0082.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0095.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0098.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0113.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0125.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0133.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0135.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0151.pdf",
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0165.pdf", # Polizeiliche Gewaltanwendung und Einsatz von Zwangsmitteln (Taser)
    "https://pardok.parlament-berlin.de/starweb/adis/citat/VT/19/DruckSachen/d19-0167.pdf"
]

pdf_texts = []
for url in urls:
    response = requests.get(url, timeout=60)
    pdf_bytes = response.content
    reader = PdfReader(BytesIO(pdf_bytes))
    pdf_text = "\n".join(page.extract_text() or "" for page in reader.pages)
    pdf_texts.append(pdf_text)
```

```python
from openai import OpenAI

# 2. Send to LLM
client = OpenAI(
    base_url="https://llm3-compute.cms.hu-berlin.de/v1/",
    api_key="required-but-not-used",
)

models = client.models.list()

chat_completion = client.chat.completions.create(
    model=models.data[0].id,
    messages=[
        {
            "role": "system",
            "content": (
                '''You are analyzing one document in isolation.

Task:
Evaluate the document using the independent metrics defined below. Treat each metric separately; do not let one score influence another.

Metric definitions:

1. Content Complexity
Conceptual difficulty, density of specialized or technical terminology, and amount of prior knowledge required beyond general legal literacy.

2. Structural Complexity
Depth of section and subsection hierarchy, frequency of internal or external cross-references, use of tables, figures, annexes, or amendments to other documents, and layout irregularity.

3. Topic Controversiality
Degree to which the document substantively addresses subjects that are known to have active, widely recognized disagreement among credible stakeholders (e.g., political parties, courts, regulators, civil society).
• Do NOT treat the existence of legal regulation alone as controversial.
• Score higher only when the text itself engages with or clearly concerns disputed policy areas.

4. Topic Diversity
Number of distinct, non-overlapping policy domains that receive substantive discussion.
• A policy domain is a broad area of public policy (e.g., education, taxation, healthcare, labor).
• Multiple aspects, instruments, or stakeholder groups within the same policy domain count as ONE topic.
• Count a topic only if it occupies a substantial portion of the document.
• Administrative, financial, or personnel provisions related to the same domain do not increase topic diversity.


Calibration rules:
• Assume a typical law proposal with standard legal language, a single main policy area, and routine structure corresponds to a score of 3 unless clear evidence supports a higher or lower score.
• Assign scores of 4 or 5 only when multiple concrete indicators for that metric are present in the text.
• Reserve a score of 5 for exceptional cases; most documents should score between 2 and 4.
• Do not treat standard legal boilerplate, definitions, or formal structure as indicators of high scores by themselves.

Topic diversity anchors:
• 1 = Single, narrowly focused policy domain
• 2 = One main domain with limited secondary aspects
• 3 = One main domain plus one clearly distinct secondary domain
• 4 = Two to three distinct, equally substantive policy domains
• 5 = Four or more distinct policy domains

Scoring scale (all metrics):
Use an integer from 1 to 5.
• 1 = Very low / minimal
• 3 = Moderate
• 5 = Very high

Rules:
• Base all scores only on the document text provided
• Do not compare to other documents
• Do not infer missing information or external context
• If uncertain, choose the lower score
• Use the metric definitions exactly as written

Output format (JSON only, no explanation):

{
  "content_complexity": 0,
  "structural_complexity": 0,
  "topic_controversiality": 0,
  "topic_diversity": 0,
  "primary_topics": []
}
'''
            ),
        },
        {
            "role": "user",
            "content": pdf_texts[9]  # <- extracted text of your PDF
        }
    ],
)

print(chat_completion.choices[0].message.content)
```
