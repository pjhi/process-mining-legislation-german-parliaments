### Event Data Embeddings

```python
import ast

BATCH_SIZE = 100
event_data_embeddings = []

events_data_concatenated = [
    f"Dieser Text beschreibt einen Gesetzesentwurf. Titel: {a['title']} "
    f"Abstrakt: {a['abstract']} "
    f"Deskriptoren: {', '.join(ast.literal_eval(a['vorgangsDeskriptoren']))}"
    for a in events_data
]

for i in range(0, len(events_data_concatenated), BATCH_SIZE):
    batch = events_data_concatenated[i:i + BATCH_SIZE]
    event_data_embeddings.extend(helper.get_embeddings(batch))

print(type(event_data_embeddings))
print(len(event_data_embeddings))
print(len(event_data_embeddings[0]))
print(event_data_embeddings[0][:10])
```

```python
import plotly.express as px
import pandas as pd
import numpy as np

# short labels
short_labels = [e["title"][:80] for e in events_data]

# PCA
X = np.array(event_data_embeddings)
X2 = PCA(n_components=2).fit_transform(X)

df = pd.DataFrame({
    "x": X2[:, 0],
    "y": X2[:, 1],
    "title": [e["title"][:200] for e in events_data]
})

fig = px.scatter(
    df,
    x="x",
    y="y",
    hover_data=["title"],
    title="Event Data Embeddings (PCA, hover for title)"
)

fig.show()
```

### Finding similar topic

```python
#wahlOMatTopics
#wahlOMatTopics_embeddings
#event_data_embeddings

E = np.array(event_data_embeddings)            # shape: (num_events, dim)
T = np.array(wahlOMatTopics_embeddings)        # shape: (num_topics, dim)

# Normalize (apparently cosine similarity works better with normalized vectors)
E = E / np.linalg.norm(E, axis=1, keepdims=True)
T = T / np.linalg.norm(T, axis=1, keepdims=True)

# Get similarity matrix
# shape: (num_events, num_topics)
# Each row = one Gesetzesentwurf
# Each column = one topic
# Each value = cosine similarity (≈ 0–1)
similarity_matrix = E @ T.T

# find closest topic per event
best_topic_indices = np.argmax(similarity_matrix, axis=1)
best_scores = np.max(similarity_matrix, axis=1)

for i, topic_idx in enumerate(best_topic_indices):
    print("Titel:", events_data[i]["title"])
    print("Bestes Thema:", wahlOMatTopics.iloc[topic_idx])
    print(f"Ähnlichkeit: {best_scores[i]:.3f}")
    print("-" * 40)


# or top-k topics per event

print("\n--------Top-K Themen pro Gesetzesentwurf:----------\n")
TOP_K = 3

top_k_indices = np.argsort(similarity_matrix, axis=1)[:, -TOP_K:][:, ::-1]

for i, indices in enumerate(top_k_indices):
    print("Titel:", events_data[i]["title"])
    for idx in indices:
        print(f"  {wahlOMatTopics.iloc[idx]} → {similarity_matrix[i, idx]:.3f}")
    print("-" * 40)
```
